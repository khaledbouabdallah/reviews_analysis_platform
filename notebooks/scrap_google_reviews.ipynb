```python
scraper = GoogleMapsReviewScraper(driver_path="Driver/chromedriver", headless=False, extra_headers=["--disable-gpu"], concat_extra=True)

for i,row in businesses.iterrows():
    try:
        if row["is_Scrapped"]:
            continue
        print(f"Scraping {row['Name']}")
        data = scraper.scrap(row["URL"])
        reviews = pd.DataFrame(data)
        reviews["source"] = row["Name"]
        save_path = f'{DATA_DIR}/{row["Name"].replace(" ","_")}.csv'
        reviews.to_csv  (save_path, encoding="utf-8")
        businesses.loc[i,"is_Scrapped"] = True
    except Exception as e:
        print(f"scraping {row['Name']} failed: ",e)
        continue
    finally:
        time.sleep(0.2)

businesses.to_csv("data/businesses.csv", encoding="utf-8", index=False)
```
